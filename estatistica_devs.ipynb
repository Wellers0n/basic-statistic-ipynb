```python
import pandas as pd
```

```python
pd.__version__
```

```python
dict_medidas = {'idade': [15, 18, 25, 25, 40, 55, 58, 60, 80], 'altura': [160, 162, 165, 168, 172, 174, 174, 174, 176]}
```

```python
dict_medidas
```

```python
df_medidas = pd.DataFrame(dict_medidas)
```

```python
df_medidas
```

# Medidas resumo

### Medidas de posição

```python
# media
df_medidas['idade'].mean()
```

```python
# mediana
df_medidas['idade'].median()
```

```python
# moda
df_medidas['idade'].mode()
```

```python
# moda
df_medidas['altura'].mode()
```

# Medidas de dispersão

```python
# variancia
df_medidas['idade'].var()
```

```python
# desvio padrao
df_medidas['idade'].std()
```

```python
# coeficiente de variacao - idade
df_medidas['idade'].std() / df_medidas['idade'].mean() * 100
```

```python
# coeficiente de variacao - altura
df_medidas['altura'].std() / df_medidas['altura'].mean() * 100
```

# Medidas de forma

```python
# assimetria da idade
df_medidas['idade'].skew()
```

```python
# assimetria da altura
df_medidas['altura'].skew()
```

```python
# curtose da idade
df_medidas['idade'].kurtosis()
```

```python
# curtose da altura
df_medidas['altura'].kurtosis()
```

```python
# descreva
df_medidas.describe()
```
